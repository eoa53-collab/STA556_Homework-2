### Does average exam scores differ across motivation levels?

```python
motivation_summary = summarize_group(data, "Motivation_Level")
motivation_summary["mean"] = motivation_summary["mean"].round(2)
motivation_summary
```

### Interpretation

Students with high motivation had the highest average exam score of 67.70, followed by students with medium motivation at 67.33 and students with low motivation at 66.73. Although the pattern suggests that greater motivation is associated with slightly higher exam performance, the difference between the high and low motivation groups is less than one exam point.

### What is the relationship between the hours students studied and their exam score?

```python
data[["Hours_Studied", "Exam_Score"]].describe().round(3)
```

```python
correlation = data[["Hours_Studied", "Exam_Score"]].corr().round(4)
correlation
```

The correlation between hours studied and exam score is 0.4465, indicating a moderate positive relationship. Students who spend more hours studying tend to have higher exam scores, although the relationship is not perfect.

### How does the Attendance of students impact their Exams scores

```python
data[["Attendance", "Exam_Score"]].describe().round(3)
```

```python
Attendance_corr = data[["Attendance", "Exam_Score"]].corr().round(4)
Attendance_corr.loc["Attendance", "Exam_Score"]
```

### Interpretation

Since there is a moderate positive correlation between the attendance of students and their final exams score, it means students who attend classes often tend to get higher scores
