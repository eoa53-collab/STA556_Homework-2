```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("../data/processed/student_performance_clean.csv")
data.head()
```

### Distribution of Exam Scores

```python
fig, ax = plt.subplots(figsize=(9, 7))

sns.histplot(data = data, x= "Exam_Score", bins= 25, edgecolor = "black")


ax.set(title=( "Figure 1 : Distribution of Student Exam Scores" ),
    xlabel="Exam Score (points)",
    ylabel="Number of Students"
)

fig.tight_layout()

plt.savefig(
    "../figures/Distribution of Student Exam Scores.png",
    bbox_inches="tight"
)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 7))

sns.kdeplot(data = data, x= "Exam_Score")


ax.set(title=( "Figure 2 : Density of Student Exam Scores" ),
    xlabel="Exam Score (points)"
)

fig.tight_layout()

plt.savefig("../figures/Density of Student Exam Scores.png",bbox_inches="tight")
plt.show()
```

Figure 1 shows the distribution of student exam scores and Figure 2 shows the Density of the student's exam scores. Most students' scores are concentrated around the middle 60s, indicating that exam performance is centered near approximately 67 points rather than being evenly distributed across the full scoring range.

### Comparison Across Motivation Groups

```python
fig, ax = plt.subplots(figsize=(9, 7))

sns.boxplot(data=data, x="Motivation_Level", y="Exam_Score", order=["Low","Medium","High"], color="white", linecolor="black", ax=ax)

#sns.stripplot(data=data, x="Motivation_Level", y="Exam_Score",alpha=0.5,jitter=True,ax=ax)


ax.set(title=( "Figure 3 : Exam Scores by Motivation Level" ),
    xlabel="Motivation Level",
    ylabel="Exam Score (points)"
)

fig.tight_layout()

plt.savefig("../figures/Exam Scores by Motivation Level.png", bbox_inches="tight")
plt.show()
```

 Figure 3 shows the Exam scores by motivation level students receive.This shows that high-motivation students have a slightly higher average score, but the differences between the groups are relatively small.

### Relationship Between Study Hours and Exam Score

```python
fig, ax = plt.subplots(figsize=(9, 7))

sns.scatterplot(data=data, x = "Hours_Studied", y= "Exam_Score",alpha = 0.7,  ax=ax)

sns.regplot(data=data, x="Hours_Studied", y="Exam_Score", scatter=False, color="purple")

ax.set(title=( "Figure 4: Hours Studied vs. Exam Score" ),
    xlabel= "Hours Studied per Week",
    ylabel="Exam Score (points)"
)

fig.tight_layout()

plt.savefig("../figures/Hours Studied and Exam Scores Relationship.png",bbox_inches="tight")
plt.show()
```

Figure 4 shows the relationship between hours studied and exams score. The upward pattern in the points indicates that students who study for more hours generally obtain higher exam scores. There is a linear relationship between the hours studied and the Exams Scores

```python

g = sns.relplot(data=data, x="Hours_Studied", y="Exam_Score",  col="Gender", hue="Gender", legend=False)

g.set_axis_labels("Hours Studied per Week", "Exam Score (points)")
g.figure.suptitle("Figure 5: Hours Studied vs. Exam Score by Gender", y=1.05)

g.figure.tight_layout()

plt.savefig("../figures/Hours Studied and Exam Scores Relationship by Gender.png", bbox_inches="tight")
plt.show()
```

Figure 5 shows the relationship between hours studied and exam score for male and female students. In both panels, the points have an upward trend, indicating that students who study more hours per week generally obtain higher exam scores. The relationship appears linear and positive, and it is similar for both genders.
